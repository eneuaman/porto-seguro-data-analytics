# 01 — Data Loading & First Inspection

**Porto Seguro's Safe Driver Prediction** (OpenML id 42742).

Scope of this notebook: load the raw data and inspect its basic structure only.
No EDA plots, no feature engineering, no modelling — those live in later notebooks.

Points established here that feed later work:
- class balance of the target,
- missing values are encoded as `-1` (not `NaN`),
- feature groups implied by the naming convention (`_bin`, `_cat`, `_calc`, remainder).

```python
import numpy as np
import pandas as pd
from sklearn.datasets import fetch_openml

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 160)
print("pandas", pd.__version__, "| numpy", np.__version__)
```

## Load the data

`fetch_openml` downloads the dataset once and caches it outside the repo (`~/scikit_learn_data`), so the raw data is never committed.

```python
bunch = fetch_openml(data_id=42742, as_frame=True)
df = bunch.frame.copy()

target_col = "target"
id_cols = [c for c in df.columns if c.lower() == "id"]
feature_cols = [c for c in df.columns if c != target_col and c not in id_cols]

if id_cols:
    print(f"Non-feature id column(s) present and excluded from features: {id_cols}")
print(f"Loaded frame: {df.shape[0]:,} rows x {df.shape[1]} columns "
      f"({len(feature_cols)} features + target"
      f"{' + id' if id_cols else ''}).")
```

## Shape, dtypes and first rows

```python
print("Shape:", df.shape)
print("\nDtype counts:")
print(df.dtypes.value_counts())
print("\nPer-column dtypes:")
print(df.dtypes.to_string())
```

## Target class balance

`target` = 1 means the customer filed a claim. Counts **and** percentages, so the class imbalance is explicit (it drives metric choice later).

```python
vc = df[target_col].value_counts().sort_index()
vp = (vc / len(df) * 100).round(4)
balance = pd.DataFrame({"count": vc, "percent": vp})
print("Target class balance (0 = no claim, 1 = claim):")
print(balance.to_string())

minority, majority = vc.min(), vc.max()
print(f"\nImbalance ratio (majority : minority) = {majority / minority:.1f} : 1")
print(f"Positive (claim) rate = {minority / len(df) * 100:.4f}%")
```

## Missing values

In the **raw Kaggle CSV**, missing values are coded as `-1`. **This OpenML load (id 42742) does not use that coding**: the ARFF declares a missing token, so `fetch_openml` already maps missing entries to `NaN`. We therefore
1. verify that **no `-1` codes remain** in this load, then
2. report the count and share of **`NaN`** per feature column.

The data is **not** treated as complete — the missingness simply lives in `NaN` here rather than in a `-1` sentinel.

```python
# (1) Confirm the raw '-1' missing code is absent in this OpenML load
#     (robust across category dtype, where a plain `== -1` would silently miss it).
neg1_total = 0
for c in feature_cols:
    s = df[c]
    if isinstance(s.dtype, pd.CategoricalDtype):
        if any(str(x).strip() in ("-1", "-1.0") for x in s.cat.categories):
            neg1_total += int((s.astype("string") == "-1").sum())
    else:
        neg1_total += int((s == -1).sum())
print(f"Total '-1' values across all {len(feature_cols)} feature columns: {neg1_total}")

# (2) Real missingness in this load is encoded as NaN -> count and % per column
nan_counts = df[feature_cols].isna().sum()
nan_perc = (nan_counts / len(df) * 100).round(4)
missing = pd.DataFrame({"nan_count": nan_counts, "nan_percent": nan_perc})
missing = missing[missing["nan_count"] > 0].sort_values("nan_count", ascending=False)

print(f"Total NaN in features: {int(nan_counts.sum()):,}")
print(f"{len(missing)} of {len(feature_cols)} feature columns contain NaN:\n")
print(missing.to_string())
```

## Feature buckets from the naming convention

Porto Seguro feature names carry type information:
- `_bin` — binary,
- `_cat` — nominal / categorical,
- `_calc` — calculated features (many Kagglers drop these),
- remainder — numeric without a `bin`/`cat` suffix, i.e. **ordinal / continuous**.

Precedence is **suffix-first** in the order `_bin -> _cat -> _calc -> remainder`, so a `calc`-group feature ending in `_bin` (e.g. `ps_calc_15_bin`) is counted under `_bin`. The total count of `calc`-named features (the superset relevant to a later “drop calc” decision) is reported separately.

```python
bin_cols  = [c for c in feature_cols if c.endswith("_bin")]
cat_cols  = [c for c in feature_cols if c.endswith("_cat")]
calc_cols = [c for c in feature_cols if ("calc" in c) and c not in bin_cols and c not in cat_cols]
remainder = [c for c in feature_cols if c not in bin_cols + cat_cols + calc_cols]

print("Feature buckets (suffix-first: _bin -> _cat -> _calc -> remainder):")
for name, cols in [("_bin (binary)", bin_cols),
                   ("_cat (nominal)", cat_cols),
                   ("_calc (calculated)", calc_cols),
                   ("remainder (ordinal/continuous)", remainder)]:
    print(f"  {name:34s}: {len(cols):2d}")
total = len(bin_cols) + len(cat_cols) + len(calc_cols) + len(remainder)
print(f"  {'TOTAL':34s}: {total:2d}  (= {len(feature_cols)} features)")

all_calc = [c for c in feature_cols if "calc" in c]
print(f"\nNote: {len(all_calc)} features are 'calc'-named in total "
      f"({len(all_calc) - len(calc_cols)} of them end in _bin/_cat and are bucketed above).")

print("\nBucket members:")
for name, cols in [("_bin", bin_cols), ("_cat", cat_cols),
                   ("_calc", calc_cols), ("remainder", remainder)]:
    print(f"\n{name} ({len(cols)}): {cols}")
```
